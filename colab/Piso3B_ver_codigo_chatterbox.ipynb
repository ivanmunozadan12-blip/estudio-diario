```python
!pip install -q -U huggingface_hub
import os
from huggingface_hub import snapshot_download, list_repo_files
SP = "ResembleAI/Chatterbox-Multilingual-TTS-es-es"
print("== ARCHIVOS DE LA DEMO ==")
try:
    for f in list_repo_files(SP, repo_type="space"): print(f)
except Exception as e: print("ERROR lista:", e)
d = snapshot_download(SP, repo_type="space", allow_patterns=["*.py", "*.txt", "*.md", "*.json", "*.yaml", "*.yml"])
for nombre in ["requirements.txt", "app.py"]:
    p = os.path.join(d, nombre)
    if os.path.exists(p):
        t = open(p, encoding="utf-8", errors="ignore").read()
        print(f"\n===== {nombre} ({len(t)} caracteres) =====\n" + t[:14000])
print("\n== ARCHIVOS DEL MODELO ==")
try:
    for f in list_repo_files("ResembleAI/Chatterbox-Multilingual-es-es"): print(f)
except Exception as e: print("ERROR lista modelo:", e)
```
